```python
#Para descobrir qual pizza vende mais e qual vende menos, preciso agrupar os dados por nome da pizza e somar a quantidade vendida de cada uma.
#1- Agrupar por pizza e somar a quantidade
vendas_por_pizza = df.groupby('pizza')['quantidade'].sum()
vendas_por_pizza


```

```python
#para encontrar a pizza que mais vende e a quantidade mais
pizza_mais_vendida = vendas_por_pizza.idxmax()
quantidade_mais_vendida = vendas_por_pizza.max()
print(pizza_mais_vendida)
print(quantidade_mais_vendida)
```

```python
#para a menos vendida
pizza_menos_vendida = vendas_por_pizza.idxmin()
quantidade_menos_vendida = vendas_por_pizza.min()
print(pizza_menos_vendida)
print(quantidade_menos_vendida)
```

```python
# Exibir os resultados
print(f"A pizza que mais vende é a '{pizza_mais_vendida}' com {quantidade_mais_vendida} unidades vendidas.")
print(f"A pizza que menos vende é a '{pizza_menos_vendida}' com {quantidade_menos_vendida} unidades vendidas.")
```

```python
# Calcular o total de vendas para cada linha (valor_unitario * quantidade)
df['total_venda'] = df['valor_unitario'] * df['quantidade']

# Calcular o total geral de vendas
total_geral_vendas = df['total_venda'].sum()

# Exibir o total geral de vendas
print(f"O total de vendas da pizzaria é: R$ {total_geral_vendas:.2f}")
```

```python
# Filtrar o DataFrame para obter apenas as vendas de pizza Portuguesa
vendas_portuguesa = df[df['pizza'] == 'Portuguesa']

# calcular o total de vendas para pizza de portuguesa
total_vendas_portuguesa = vendas_portuguesa['total_venda'].sum()

# Exibir o resultado
print(f"Com a venda de pizzas de Portuguesa, você ganhou R$ {total_vendas_portuguesa:.2f}")
```
